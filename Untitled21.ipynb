### 1.1 Captura de Imágenes desde la Cámara del PC y Almacenamiento en MongoDB

Para capturar imágenes directamente desde la cámara de tu PC, utilizaremos `opencv-python`. Ten en cuenta que en entornos como Google Colab, acceder directamente a la cámara del usuario requiere un poco de interactividad y el navegador pedirá permiso.

El siguiente código te permitirá:
1.  **Capturar una imagen** de tu webcam.
2.  **Preprocesar**la (convertir a escala de grises, redimensionar a 10x10 píxeles, aplanar).
3.  **Asignarle una etiqueta** (ej. 'feliz', 'triste', 'enojado', 'preocupado').
4.  **Guardar** la imagen preprocesada y su etiqueta en la colección `image_classification_data` en MongoDB.

Necesitarás ejecutar la celda de captura varias veces, una para cada expresión que quieras entrenar, asegurándote de cambiar la etiqueta (`current_label`) y de que tu rostro esté visible en la cámara al momento de la captura.

```python
def take_photo(filename='photo.jpg', quality=0.8):
    js = Javascript('''
        async function takePhoto(quality) {
            const div = document.createElement('div');
            const capture = document.createElement('button');
            capture.textContent = 'Capturar Imagen';
            div.appendChild(capture);

            const video = document.createElement('video');
            video.style.display = 'block';
            const stream = await navigator.mediaDevices.getUserMedia({video: true});

            document.body.appendChild(div);
            div.appendChild(video);
            video.srcObject = stream;
            await video.play();

            // Redimensionar la imagen a un tamaño más pequeño para evitar sobrecargar Colab
            // Esto es solo para la vista previa, la imagen final se ajustará más tarde.
            // google.colab.output.set_dimensions(video.videoWidth, video.videoHeight); // Comentado para evitar TypeError

            await new Promise((resolve) => capture.onclick = resolve);

            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);
            stream.getVideoTracks()[0].stop();
            div.remove();
            return canvas.toDataURL('image/jpeg', quality);
        }
    ''')
    display(js)
    data = eval_js('takePhoto({})'.format(quality))
    binary = base64.b64decode(data.split(',')[1])
    with open(filename, 'wb') as f:
        f.write(binary)
    return filename

# --- Función para capturar, preprocesar y guardar en MongoDB ---
def capture_and_store_image(label, img_size=(10, 10)):
    print(f"Preparado para capturar una imagen para la etiqueta: '{label}'")
    try:
        filename = take_photo()
        print('Foto capturada!')

        # Leer la imagen capturada con OpenCV
        img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
        if img is None:
            print("Error: No se pudo cargar la imagen. Asegúrate de que la cámara funcione y se haya capturado una foto.")
            return

        # Redimensionar la imagen a 10x10 píxeles
        img_resized = cv2.resize(img, img_size, interpolation=cv2.INTER_AREA)

        # Normalizar y aplanar la imagen
        img_flattened = img_resized.flatten().astype(np.float32) # Convertir a float32
        img_normalized = img_flattened / 255.0 # Normalizar a 0-1

        # Mostrar la imagen preprocesada (opcional, para verificación)
        # display(Image(data=cv2.imencode('.png', img_resized)[1].tobytes()))
        # print(f"Imagen preprocesada para '{label}'. Shape: {img_resized.shape}")

        # Guardar en MongoDB
        if client is not None and db is not None:
            image_collection = db.image_classification_data
            try:
                image_collection.insert_one({"image": img_normalized.tolist(), "label": label})
                print(f"Imagen preprocesada para '{label}' guardada en MongoDB.")
            except Exception as e:
                print(f"Error al guardar la imagen en MongoDB: {e}")
        else:
            print("No se pudo guardar la imagen: el cliente o la base de datos de MongoDB no están disponibles.")

    except Exception as err:
        print(str(err))
        print("Error al capturar o procesar la foto. Asegúrate de otorgar permisos a la cámara.")

print("Función 'capture_and_store_image' definida.")
```

### 2. Carga y Preprocesamiento de Datos de Imágenes

Ahora cargaremos los datos de imágenes desde MongoDB y realizaremos un preprocesamiento básico.

```python
# Cargar los datos de imágenes desde MongoDB
if client is not None and db is not None:
    image_collection = db.image_classification_data
    loaded_image_data = list(image_collection.find({}))

    if loaded_image_data:
        df_images = pd.DataFrame(loaded_image_data)
        if '_id' in df_images.columns:
            df_images = df_images.drop(columns=['_id'])
        print("Datos de imágenes cargados desde 'image_classification_data'.")
    else:
        print("Advertencia: La colección 'image_classification_data' está vacía.")
        df_images = pd.DataFrame()
else:
    print("Advertencia: La conexión a MongoDB no está disponible. No se pudieron cargar los datos de imágenes.")
    df_images = pd.DataFrame()

display(df_images.head())

# Preprocesamiento: Normalización de píxeles y separación de características y etiquetas
if not df_images.empty:
    X_images = np.array(df_images['image'].tolist()) # Convertir la lista de píxeles a array NumPy
    y_images = df_images['label']

    # Normalización: Escalar los valores de los píxeles a un rango de 0 a 1
    X_images = X_images / X_images.max() if X_images.max() > 0 else X_images # Evitar división por cero

    # Codificar etiquetas categóricas (feliz/triste) a numéricas (0/1)
    from sklearn.preprocessing import LabelEncoder
    label_encoder = LabelEncoder()
    y_images_encoded = label_encoder.fit_transform(y_images)
    print(f"Clases de etiquetas: {label_encoder.classes_}")

    print("Datos de imágenes preprocesados y etiquetas codificadas.")
else:
    print("No hay datos para preprocesar.")
```

### 3. Partición del Dataset de Imágenes

Dividiremos los datos preprocesados en conjuntos de entrenamiento y prueba.

```python
from sklearn.model_selection import train_test_split

if 'X_images' in locals() and not df_images.empty:
    X_train_img, X_test_img, y_train_img, y_test_img = train_test_split(
        X_images, y_images_encoded, test_size=0.2, random_state=42, stratify=y_images_encoded
    )

    print(f"Tamaño del conjunto de entrenamiento de imágenes: {len(X_train_img)} muestras")
    print(f"Tamaño del conjunto de prueba de imágenes: {len(X_test_img)} muestras")

    print("Primeras 5 muestras de X_train_img (planas):")
    print(X_train_img[:5])
    print("Primeras 5 muestras de y_train_img:")
    print(y_train_img[:5])
else:
    print("No hay datos de imágenes disponibles para la partición.")
```

### 4. Modelo y Ajuste para Clasificación de Imágenes (CNN)

### 4.1 Modelo de Clasificación de Imágenes con TensorFlow y Keras

Vamos a usar una red neuronal convolucional (CNN) simple con TensorFlow y Keras para la clasificación de expresiones faciales. Este enfoque es más adecuado para datos de imágenes que los modelos lineales como la Regresión Logística, ya que las CNNs pueden aprender jerarquías de características espaciales.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np # Ensure numpy is imported for reshaping

# Check if X_train_img and X_test_img are defined
if 'X_train_img' in locals() and 'X_test_img' in locals():
    # Reshape the images for CNN input (10x10x1 for grayscale images)
    # Assuming img_size was (10, 10) in capture_and_store_image
    img_height, img_width = 10, 10
    X_train_img_reshaped = X_train_img.reshape(-1, img_height, img_width, 1)
    X_test_img_reshaped = X_test_img.reshape(-1, img_height, img_width, 1)

    # Define the Keras model
    keras_model = keras.Sequential([
        layers.Input(shape=(img_height, img_width, 1)),
        layers.Conv2D(32, kernel_size=(3, 3), activation='relu'),
        layers.MaxPooling2D(pool_size=(2, 2)),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dense(len(label_encoder.classes_), activation='softmax') # Output layer for classification
    ])

    # Compile the model
    keras_model.compile(optimizer='adam',
                        loss='sparse_categorical_crossentropy',
                        metrics=['accuracy'])

    keras_model.summary()
    print("Modelo de Keras definido y compilado.")
else:
    print("Error: X_train_img o X_test_img no están definidos. Por favor, asegúrate de ejecutar las celdas de carga y partición de datos (Celdas 2 y 3) antes de esta celda.")
```

```python
# Removed Logistic Regression model definition as CNN is now used.
```

```python
if 'keras_model' in locals() and 'X_train_img_reshaped' in locals():
    # Train the Keras model
    history = keras_model.fit(X_train_img_reshaped, y_train_img,
                              epochs=10, # You can adjust the number of epochs
                              batch_size=32,
                              validation_data=(X_test_img_reshaped, y_test_img))
    print("Modelo de clasificación de imágenes (Keras CNN) entrenado exitosamente.")
else:
    print("No hay modelo de Keras o datos de entrenamiento disponibles para entrenar.")
```

### 5. Evaluación del Modelo de Clasificación de Imágenes

Evaluaremos el rendimiento del modelo utilizando métricas de clasificación.

```python
# Removed Logistic Regression model evaluation as CNN is now used. See cell 4421b2a9 for Keras CNN evaluation.
```

```python
if 'keras_model' in locals() and 'X_test_img_reshaped' in locals():
    # Evaluate the Keras model
    loss, accuracy_img = keras_model.evaluate(X_test_img_reshaped, y_test_img, verbose=0)
    print(f"Precisión del modelo de imágenes (Keras CNN): {accuracy_img:.2f}")

    y_pred_probs = keras_model.predict(X_test_img_reshaped)
    y_pred_img = np.argmax(y_pred_probs, axis=1)

    print("\nReporte de Clasificación del Modelo de Imágenes (Keras CNN):")
    print(classification_report(y_test_img, y_pred_img, target_names=label_encoder.classes_))

    conf_matrix_img = confusion_matrix(y_test_img, y_pred_img)
    print("\nMatriz de Confusión del Modelo de Imágenes (Keras CNN):")
    print(conf_matrix_img)

    plt.figure(figsize=(6, 4))
    sns.heatmap(conf_matrix_img, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=label_encoder.classes_, yticklabels=label_encoder.classes_)
    plt.xlabel('Predicción')
    plt.ylabel('Valor Real')
    plt.title('Matriz de Confusión de Clasificación de Imágenes (Keras CNN)')
    plt.show()
else:
    print("No se pudo evaluar el modelo: el modelo o los datos de prueba no están disponibles.")

print("\nFlujo de trabajo ML de clasificación de imágenes completado.")
```
